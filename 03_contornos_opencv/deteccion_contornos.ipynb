```python
import numpy as np
import cv2

# --- Crear imagen sintética ---
height, width = 400, 400
image = np.zeros((height, width), dtype=np.uint8)

# Dibujar figuras geométricas
cv2.rectangle(image, (50, 50), (150, 200), 150, -1)      # Rectángulo (gris claro)
cv2.circle(image, (300, 100), 50, 220, -1)               # Círculo (casi blanco)
pts = np.array([[250, 300], [350, 350], [300, 250]], np.int32)  # Triángulo
pts = pts.reshape((-1, 1, 2))
cv2.fillPoly(image, [pts], 100)                          # Triángulo (gris oscuro)

# ---  Histograma de la imagen sintética ---
hist = cv2.calcHist([image], [0], None, [256], [0, 256])
print("\nHistograma de intensidades (primeros 20 valores):\n", hist[:20].flatten())

# ---  Umbralizar la imagen ---
ret, thresh = cv2.threshold(image, 10, 255, cv2.THRESH_BINARY)

# Histograma de la imagen umbralizada
hist_thresh = cv2.calcHist([thresh], [0], None, [256], [0, 256])
print("\nHistograma de imagen umbralizada (primeros 20 valores):\n", hist_thresh[:20].flatten())

# ---  Detección de contornos ---
contours_none, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
image_none = cv2.cvtColor(image.copy(), cv2.COLOR_GRAY2BGR)
cv2.drawContours(image_none, contours_none, -1, (0, 0, 255), 2)

contours_simple, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
image_simple = cv2.cvtColor(image.copy(), cv2.COLOR_GRAY2BGR)
cv2.drawContours(image_simple, contours_simple, -1, (0, 255, 0), 2)

# Visualización de los puntos individuales del primer contorno (NONE)
img_points = cv2.cvtColor(image.copy(), cv2.COLOR_GRAY2BGR)
for pt in contours_none[0]:
    cv2.circle(img_points, tuple(pt[0]), 1, (255,0,0), -1)

# ---  Estadísticas en consola ---
print("\n--- ÁREAS Y PERÍMETROS DE LOS CONTORNOS (CHAIN_APPROX_SIMPLE) ---")
for idx, cnt in enumerate(contours_simple):
    area = cv2.contourArea(cnt)
    perimeter = cv2.arcLength(cnt, True)
    print(f"Contorno {idx+1} (SIMPLE): Área = {area:.2f}, Perímetro = {perimeter:.2f}, Puntos = {len(cnt)}")

print("\n--- ÁREAS Y PERÍMETROS DE LOS CONTORNOS (CHAIN_APPROX_NONE) ---")
for idx, cnt in enumerate(contours_none):
    area = cv2.contourArea(cnt)
    perimeter = cv2.arcLength(cnt, True)
    print(f"Contorno {idx+1} (NONE): Área = {area:.2f}, Perímetro = {perimeter:.2f}, Puntos = {len(cnt)}")

print(f"\nNúmero de puntos primer contorno (NONE): {len(contours_none[0])}")
print(f"Número de puntos primer contorno (SIMPLE): {len(contours_simple[0])}")
print("Ejemplo de puntos (NONE):", contours_none[0][:5].squeeze())
print("Ejemplo de puntos (SIMPLE):", contours_simple[0][:5].squeeze())

# ---  Mostrar todas las imágenes a la vez y posicionarlas ---
cv2.imshow("Imagen sintética", image)
cv2.moveWindow("Imagen sintética", 0, 0)

cv2.imshow("Imagen umbralizada", thresh)
cv2.moveWindow("Imagen umbralizada", 450, 0)

cv2.imshow("Contornos CHAIN_APPROX_NONE", image_none)
cv2.moveWindow("Contornos CHAIN_APPROX_NONE", 0, 450)

cv2.imshow("Contornos CHAIN_APPROX_SIMPLE", image_simple)
cv2.moveWindow("Contornos CHAIN_APPROX_SIMPLE", 450, 450)

cv2.imshow("Puntos del primer contorno (CHAIN_APPROX_NONE)", img_points)
cv2.moveWindow("Puntos del primer contorno (CHAIN_APPROX_NONE)", 900, 0)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
